```python
import sympy as sp
import pandas as pd
import numpy as np
from IPython.display import display, HTML, Math

def newton_raphson_ejemplo28(x0_val=1000.0, tol=1e-6, max_iter=100):
    # 1. Variable simbólica
    x = sp.Symbol('x')

    # 2. Definición de la ecuación de balance térmico
    f_expr = (9.69779e7
              - 49205.745 * (x - 298)
              - (6.48421 / 2) * (x**2 - 298**2)
              - (0.0079433 / 3) * (x**3 - 298**3)
              + (3.46313e-6 / 4) * (x**4 - 298**4))

    df_expr = sp.diff(f_expr, x)

    # Encabezado visual
    html_header = """
    <div style="background-color: #f8f9fa; border-left: 5px solid #007bff; padding: 15px; margin-bottom: 15px; border-radius: 4px;">
        <h2 style="color: #007bff; margin-top: 0; font-family: sans-serif;">Método de Newton-Raphson</h2>
        <p style="font-size: 15px; font-family: sans-serif; color: #333;">
            Ejemplo 28: Balance Térmico
        </p>
    </div>
    """
    display(HTML(html_header))

    display(Math(rf"\mathbf{{Función\ evaluada:}}\quad f(x) = 9.69779 \times 10^7 - 49205.745(x-298) - 3.242105(x^2 - 298^2) - 0.00264777(x^3 - 298^3) + 8.657825 \times 10^{{-7}}(x^4 - 298^4) = 0"))
    display(Math(rf"\mathbf{{Derivada\ analítica:}}\quad f'(x) = -49205.745 - 6.48421 x - 0.0079433 x^2 + 3.46313 \times 10^{{-6}} x^3"))

    # Conversión a funciones numéricas
    f_num = sp.lambdify(x, f_expr, modules=['numpy'])
    df_num = sp.lambdify(x, df_expr, modules=['numpy'])

    # Algoritmo de Newton-Raphson
    iteraciones = []
    x_n = float(x0_val)

    for n in range(max_iter):
        fx = float(f_num(x_n))
        dfx = float(df_num(x_n))

        error = np.nan if n == 0 else abs(x_n - x_prev)

        iteraciones.append({
            'n': n,
            'x_n': x_n,
            'f(x_n)': fx,
            'f\'(x_n)': dfx,
            'Error': error
        })

        if n > 0 and error < tol:
            break

        x_prev = x_n
        x_n = x_n - (fx / dfx)

    df_res = pd.DataFrame(iteraciones)

    # Formato visual de la tabla
    df_res_styled = df_res.copy()
    df_res_styled.columns = [
        '<i>n</i>',
        '<i>x<sub>n</sub></i>',
        '<i>f(x<sub>n</sub>)</i>',
        '<i>f \'(x<sub>n</sub>)</i>',
        '<i>Error (E<sub>n</sub>)</i>'
    ]

    tabla_styled = df_res_styled.style.format({
        '<i>x<sub>n</sub></i>': '{:.6f}',
        '<i>f(x<sub>n</sub>)</i>': '{:.6e}',
        '<i>f \'(x<sub>n</sub>)</i>': '{:.6e}',
        '<i>Error (E<sub>n</sub>)</i>': lambda v: '—' if np.isnan(v) else f'{v:.6f}'
    }).set_table_styles([
        {'selector': 'th', 'props': [
            ('background-color', '#007bff'),
            ('color', '#ffffff'),
            ('font-family', 'sans-serif'),
            ('font-size', '14px'),
            ('text-align', 'center'),
            ('padding', '10px')
        ]},
        {'selector': 'td', 'props': [
            ('text-align', 'center'),
            ('font-family', 'monospace'),
            ('font-size', '13px'),
            ('padding', '8px')
        ]}
    ]).hide(axis='index')

    display(HTML("<h3 style='font-family: sans-serif; color: #333;'>Tabla de Iteraciones</h3>"))
    display(tabla_styled)

    raiz_final = df_res.iloc[-1]['x_n']
    num_iter = int(df_res.iloc[-1]['n'])

    display(HTML(f"""
    <div style="background-color: #d4edda; color: #155724; padding: 12px; border-radius: 5px; margin-top: 15px; border: 1px solid #c3e6cb; font-family: sans-serif;">
        <strong>Convergencia alcanzada:</strong> Solución <i>x</i> &approx; <code>{raiz_final:.6f}</code> (&approx; <code>{raiz_final:.2f}</code>) en <strong>{num_iter}</strong> iteraciones.
    </div>
    """))

# Ejecución en Colab
newton_raphson_ejemplo28(x0_val=1000.0, tol=1e-6)
```
