```python
import sympy as sp
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, HTML, Math

def newton_raphson_ejemplo17(x0_val=2.0, tol=1e-6, max_iter=100):
    # 1. Variable simbólica
    x = sp.Symbol('x', positive=True)

    # 2. Parámetros físicos
    e = 0.000046
    D = 0.5
    nu = 1.02e-6

    # Construcción exacta de f(x) con Logaritmo en base 10 explícito
    t1 = 0.05886 / (x**2)
    arg_log = (e / (3.7 * D)) + (5.74 * (nu**0.9)) / ((x**0.9) * (D**0.9))
    t2 = 0.25 / (sp.log(arg_log, 10)**2)

    f_expr = t1 - t2
    df_expr = sp.diff(f_expr, x)

    # Encabezado visual
    html_header = """
    <div style="background-color: #f8f9fa; border-left: 5px solid #007bff; padding: 15px; margin-bottom: 15px; border-radius: 4px;">
        <h2 style="color: #007bff; margin-top: 0; font-family: sans-serif;">Método de Newton-Raphson</h2>
        <p style="font-size: 15px; font-family: sans-serif; color: #333;">
            Ejemplo 17: Velocidad de Flujo en Tuberías (Swamee-Jain)
        </p>
    </div>
    """
    display(HTML(html_header))

    display(Math(rf"\mathbf{{Función\ evaluada:}}\quad f(x) = \frac{{0.05886}}{{x^2}} - \frac{{0.25}}{{\left[\log_{{10}}\left(\frac{{0.000046}}{{3.7(0.5)}} + \frac{{5.74 (1.02\times 10^{{-6}})^{{0.9}}}}{{x^{{0.9}} (0.5)^{{0.9}}}}\right)\right]^2}} = 0"))

    # Conversión a funciones ejecutables numéricamente
    f_num = sp.lambdify(x, f_expr, modules=['numpy'])
    df_num = sp.lambdify(x, df_expr, modules=['numpy'])

    # Algoritmo de Newton-Raphson
    iteraciones = []
    x_n = float(x0_val)

    for n in range(max_iter):
        fx = float(f_num(x_n))
        dfx = float(df_num(x_n))

        error = np.nan if n == 0 else abs(x_n - x_prev)

        iteraciones.append({
            'n': n,
            'x_n': x_n,
            'f(x_n)': fx,
            'f\'(x_n)': dfx,
            'Error': error
        })

        if n > 0 and error < tol:
            break

        x_prev = x_n
        x_n = x_n - (fx / dfx)

    df_res = pd.DataFrame(iteraciones)

    # Formato visual idéntico al LaTeX
    df_res_styled = df_res.copy()
    df_res_styled.columns = [
        '<i>n</i>',
        '<i>x<sub>n</sub></i>',
        '<i>f(x<sub>n</sub>)</i>',
        '<i>f \'(x<sub>n</sub>)</i>',
        '<i>Error (E<sub>n</sub>)</i>'
    ]

    tabla_styled = df_res_styled.style.format({
        '<i>x<sub>n</sub></i>': '{:.6f}',
        '<i>f(x<sub>n</sub>)</i>': '{:.6e}',
        '<i>f \'(x<sub>n</sub>)</i>': '{:.6e}',
        '<i>Error (E<sub>n</sub>)</i>': lambda v: '—' if np.isnan(v) else f'{v:.6f}'
    }).set_table_styles([
        {'selector': 'th', 'props': [
            ('background-color', '#007bff'),
            ('color', '#ffffff'),
            ('font-family', 'sans-serif'),
            ('font-size', '14px'),
            ('text-align', 'center'),
            ('padding', '10px')
        ]},
        {'selector': 'td', 'props': [
            ('text-align', 'center'),
            ('font-family', 'monospace'),
            ('font-size', '13px'),
            ('padding', '8px')
        ]}
    ]).hide(axis='index')

    display(HTML("<h3 style='font-family: sans-serif; color: #333;'>Tabla de Iteraciones</h3>"))
    display(tabla_styled)

    raiz_final = df_res.iloc[-1]['x_n']
    num_iter = int(df_res.iloc[-1]['n'])

    display(HTML(f"""
    <div style="background-color: #d4edda; color: #155724; padding: 12px; border-radius: 5px; margin-top: 15px; border: 1px solid #c3e6cb; font-family: sans-serif;">
        <strong>Convergencia alcanzada:</strong> Solución <i>x</i> &approx; <code>{raiz_final:.5f}</code> en <strong>{num_iter}</strong> iteraciones.
    </div>
    """))

# Ejecución en Colab
newton_raphson_ejemplo17(x0_val=2.0, tol=1e-6)
```
